## 3. Datos

`:=` da un valor numérico con unidades. `=` define una expresión simbólica.


```python
%%eng
## Datos

L := 6*m
q := 10*kN/m
E := 200*GPa
I_z := 80e6*mm**4
```

## 4. Reacciones


```python
%%eng
## Reacciones

eqFy = eq(R_A + R_B, q*L)
eqMA = eq(R_B*L, q*L*L/2)
solve(eqFy, eqMA, R_A, R_B)
```

## 5. Ley de momentos y su diagrama


```python
%%eng
## Ley de momentos

M(x) = R_A*x - q*x^2/2
numeric(subs(M(x), x, L/2), kN*m)

plot(M(x), x, 0, L)
```

## 6. Dónde se supera el momento admisible

La desigualdad se resuelve sobre un dominio: es de donde la variable saca su
unidad, y para una viga es sencillamente la viga.


```python
%%eng
solve(M(x) > 20*kN*m, x, 0, L)
```

## 7. Flecha, con resumen al final


```python
%%eng
## Flecha

d_max = 5*q*L^4/(384*E*I_z)
report(d_max)
d_adm = L/300
report(d_adm)

## Resumen

summary()
```
